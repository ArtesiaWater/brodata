# Groundwater monitoring network (GMN)

In this example we will download all the metadadata of the wells of the Groundwater Monitoring Network of Oudewater (bro-id: GMN000000003996).

```python
import brodata
```

First we request the GroundwaterMonitoringNetwork-object

```python
gmn = brodata.gmn.GroundwaterMonitoringNetwork.from_bro_id("GMN000000003996")
```

This attribute `measuringPoint` contains the bro-ids of all Groundwater Monitoring Wells that are part of this network.

```python
gmn.measuringPoint
```

The gmn-object also contains some metadata, for example the name of the network and the monitoring-purpose.

```python
gmn_data = gmn.to_dict()
gmn_data.pop("measuringPoint")
gmn_data
```

If we want to have more detailed information about the Groundwater Monitoring Wells, we have to download them individually using the method `brodata.gmw.get_data_for_bro_ids` (which takes some time). This returns a dictionary with the bro-ids as keys and the GroundwaterMonitoringWell-objects as values.

```python
gmw_ids = gmn.measuringPoint.index.levels[0].unique()
gmws = brodata.gmw.get_data_for_bro_ids(gmw_ids)
gmws
```

We want to plot them on a map, so we transform this dictionary to a GeoDataFrame using the method `brodata.bro.objects_to_gdf`.

```python
gdf = brodata.bro.objects_to_gdf(gmws)
gdf
```

Finally, we can plot these wells on an interactive map.

```python
import folium
tooltip = ["broId","wellCode", "groundLevelPosition", "wellConstructionDate", "numberOfMonitoringTubes"]
m = gdf.explore(tooltip=tooltip)
m
```
