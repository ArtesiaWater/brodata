## Multiple objects
Download all Groundwater Utilisation Facilities near Utrecht.

```python
# Extent for Utrecht region (approximate coordinates)
extent = [125000, 145000, 445000, 465000]

# download guf-characteristics
gdf=brodata.guf.get_characteristics(extent=extent)

gdf
```

Plot the column `realisedInstallation` on a map using matplotlib.

```python
f, ax = plt.subplots()
ax.axis("scaled")
ax.axis(extent)
ax = gdf.plot(
        column="realisedInstallation",
        legend=True,
        alpha=0.7,
        edgecolor="k",
        ax=ax
    )
```

Plot the column `realisedInstallation` on a map using folium. You can click on a location to view the details.

```python
import folium
# use explode to separate multi-point geometries into single points
# otherwise the popup does not work properly
m = gdf.explode().explore(
    column="realisedInstallation",
    marker_kwds={"radius":5},
    tooltip=False,
    popup=True,
    tiles=None
)

# add a background layer with 50% opacity
folium.TileLayer(
    tiles="OpenStreetMap",
    opacity=0.5,  # 0 = invisible, 1 = opaque
    name="Background"
).add_to(m)
m
```
